# VQE Demo 1: Single-Qubit Solver

This notebook demonstrates the VQE algorithm on a simple single-qubit
Hamiltonian:

$$H = -Z = \begin{pmatrix}-1 & 0 \\ 0 & 1\end{pmatrix}$$

The ground state is $|0\rangle$ with energy $-1$.
The excited state is $|1\rangle$ with energy $+1$.

```python
# Define the Hamiltonian (negative sign so |0⟩ is the ground state)
H = PauliOperator({"Z0": -1.0})
print(f"Hamiltonian: {H}")
```

```python
# Create VQE solver with hardware-efficient ansatz (depth=2)
vqe = VQE(H, n_qubits=1, ansatz_depth=2)
print(f"Number of qubits: {vqe.n_qubits}")
print(f"Number of parameters: {vqe.n_params}")
```

```python
# Check energy at identity parameters: |0⟩ → ⟨0|-Z|0⟩ = -1
energy_initial = vqe.calculate_energy([0.0, 0.0, 0.0, 0.0])
print(f"Energy at [0,0,0,0]: {energy_initial:.6f} (expect -1.0)")
```

```python
# Run VQE optimization
ground_energy = vqe.run(optimizer='SLSQP', maxiter=100)
print(f"\n=== Results ===")
print(f"Ground-state energy: {ground_energy:.6f}")
print(f"Target (exact):      -1.0")
print(f"Optimal parameters:  {np.round(vqe.optimal_params, 4)}")
print(f"Iterations:          {len(vqe.energy_history)}")
```

```python
# Plot convergence
import matplotlib.pyplot as plt
plt.plot(vqe.energy_history, '-o', markersize=3)
plt.axhline(y=-1.0, color='r', linestyle='--', label='Target (-1.0)')
plt.xlabel('Optimizer iteration')
plt.ylabel('Energy')
plt.title('VQE Convergence (1-qubit, H = -Z)')
plt.legend()
plt.grid(True)
plt.show()
```
