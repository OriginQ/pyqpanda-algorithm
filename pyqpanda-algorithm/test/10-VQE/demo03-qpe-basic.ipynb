# QPE Demo 1: Basic Phase Estimation

This notebook demonstrates QPE with the Pauli-Z operator.

- Z|0⟩ = |0⟩ = e^(2πi·0)|0⟩  →  φ = 0
- Z|1⟩ = -|1⟩ = e^(2πi·0.5)|1⟩  →  φ = 0.5

```python
# Define Z unitary
def u_z(tq):
    cir = QCircuit()
    cir << Z(tq[0])
    return cir
```

```python
# QPE on |0⟩ (default) — should give φ = 0
qpe = QPE(unitary=u_z, n_count=4, n_target=1)
phi = qpe.run()
print(f"Phase of Z on |0⟩: φ = {phi:.4f} (expect 0.0)")
```

```python
# QPE on |1⟩ — should give φ = 0.5
def prep_one(tq):
    cir = QCircuit()
    cir << X(tq[0])
    return cir

qpe2 = QPE(unitary=u_z, n_count=4, n_target=1, state_prep=prep_one)
phi2 = qpe2.run()
print(f"Phase of Z on |1⟩: φ = {phi2:.4f} (expect 0.5)")
```

```python
# Show full probability distribution (4 counting qubits → 16 bins)
dist = qpe2.run_dict()
for k, v in sorted(dist.items()):
    bar = '█' * int(v * 50)
    print(f"  |{k}⟩  {v:.4f}  {bar}")
```
